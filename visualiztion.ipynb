```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('df.csv')
df['chance'] = (df['Percentage chance of reuse or recycle']/100)*df['Tonnage']
df['no_chance'] = df['Tonnage']- df['chance']
```

```python
labels = df['Material'].tolist()
men_means = df['chance'].tolist()
women_means = df['no_chance'].tolist()
width = 0.35       # the width of the bars: can also be len(x) sequence

fig, ax = plt.subplots()

ax.bar(labels, men_means, width, label='Reuse')
ax.bar(labels, women_means, width, bottom=men_means,
       label='Waste')

ax.set_ylabel('Tonnage')
plt.xticks(rotation=90)
ax.legend()

plt.show()
```
