##### Process:
Restriction $\Rightarrow$ Type $\Rightarrow$ Capacity $\Rightarrow$ Loads $\Rightarrow$ Primary Cost

```python
import pandas as pd
```

```python
r = pd.read_csv('restrictions.csv')
vc = pd.read_csv('vehicle_costings_by material).csv')
```

```python
ptc = pd.read_excel('shell.xlsx')
```

```python
c_dict = r.drop('type', axis=1)
t_dict = r.drop('capacity t', axis=1)
```

```python
c_dict = c_dict.set_index(keys='Restrictions').T
t_dict = t_dict.set_index(keys='Restrictions').T
```

```python
c_dict = c_dict.to_dict('list')
t_dict = t_dict.to_dict('list')
```

```python
ptc["v_type"] = ptc["Restrictions"].apply(lambda x: t_dict.get(x))
```

```python
ptc["capacity"] = ptc["Restrictions"].apply(lambda x: c_dict.get(x))
```

```python
ptc['v_type'] = ptc['v_type'].str.get(0)
ptc['capacity'] = ptc['capacity'].str.get(0)
```

```python
ptc['capacity'] = pd.to_numeric(ptc['capacity'])
```

```python
vc.set_index(keys='Vehicle sizes', inplace=True)
```

```python
ptc['Cost'] = ptc.apply(lambda row: vc.loc[row.v_type, row.Type],axis=1)
```
