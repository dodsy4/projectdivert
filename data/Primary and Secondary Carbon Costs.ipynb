##### Process:

Vehicle $\Rightarrow$ Carbon per mile $\Rightarrow$ Loads $\Rightarrow$ Primary Carbon / Mile

```python
import pandas as pd
```

```python
df = pd.read_excel('primary_carbon_vt.xlsx')
ptc = pd.read_csv('mptc')
```

```python
df.set_index(keys='vehicle_sizes', inplace=True)
```

```python
carbon_dict = df.T.to_dict('list')
```

```python
ptc['pcc'] = ptc["v_type"].apply(lambda x: carbon_dict.get(x))
```

```python
ptc['pcc'] = ptc['pcc'].str.get(0)
```
